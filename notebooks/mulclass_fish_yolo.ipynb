```python
from ultralytics import YOLO

model=YOLO('yolo11n-cls') # yolo11n-cls is a pretrained model for classification
```

```python
# Train the model
model.train(data="/content/images.cv_jzk6llhf18tm3k0kyttxz/data", epochs=10 , batch=16)
```

```python
result=model.predict('/content/shrimp1.jpg')
import matplotlib.pyplot as plt
plt.imshow(result[0].plot())
plt.show()
```
